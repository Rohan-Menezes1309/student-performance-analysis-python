```python
import pandas as pd
df= pd.read_csv("student_cleaned.csv")
df.head()
```

```python
df["G3"].describe()
```

```python
df.groupby("studytime")["G3"].mean()
```

Students who studied between 3-4 hours daily achieved the highest average final grades, suggesting that increased study time generally improves performance.

```python
df.groupby("failures")["G3"].mean()
```

Students with no past academic failures achieved significantly higher final grades compared to students with one or more failures. Performance decreases consistently as the number of failures increases.

```python
df.groupby("absences")["G3"].mean()
```

```python
df.groupby("internet")["G3"].mean()
```

Students with internet access scored higher on average compared to those without internet access, suggesting that access to online resources may support academic performance.

```python
df["G3"].mean()
```

The average of final grade (G3) across all students is approximately 10.4, indicating moderate academic performance overall.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.histplot(df["G3"], bins=10)
plt.title("Distribution of Final Grades (G3)")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")
plt.savefig("distribution_of_grades.png")
plt.show()
```

```python
sns.barplot(x="studytime", y="G3", data=df)
plt.title("Study Time v/s Final Grade")
plt.xlabel("Study Time Level")
plt.ylabel("Average Final Grade")
plt.savefig("studytime_vs_grades.png")
plt.show()
```

```python
sns.barplot(x="failures", y="G3", data=df)
plt.title("Failures v/s Final Grade")
plt.xlabel("Failures")
plt.ylabel("Average Final Grade")
plt.savefig("failures_vs_grades.png")
plt.show()
```

The visualization shows a clear decline in final grades as the number of past failures increases. Students with no failures performed significantly better than those with multiple failures.

```python
sns.barplot(x="internet", y="G3", data=df)
plt.title("Internet Access v/s Final Grade")
plt.ylabel("Average Final Grade")
plt.xlabel("Internet Access")
plt.savefig("internet_access_vs_grade")
plt.show()
```

Students with internet access achieved slightly higher average final grades compared to students without internet access.

```python
sns.scatterplot(x="absences", y="G3", data=df)
plt.title("Absences v/s Average Final Grade")
plt.xlabel("Absences")
plt.ylabel("Average Final Grade")
plt.savefig("absences_vs_grade")
plt.show()
```

The scatter plot indicates that grades remain relatively stable for students with fewer than 15 absences, but performance begins to decline as absences increase beyond this level.
